## 5.1 Les Types de Base

Python est un langage **dynamiquement typé** : on ne déclare pas le type d'une variable, l'interpréteur le déduit automatiquement.

| Type | Description | Exemple | Mutable ? |
| --- | --- | --- | --- |
| `int` | Entier (integer en anglais) | `42`, `-7`, `1_000_000` | ❌ Non |
| `float` | Nombre à virgule flottante (64 bits, float) | `3.14`, `-0.001`, `1e6` | ❌ Non |
| `bool` | Boolean | `True`, `False` | ❌ Non |
| `str` | Chaîne de caractères (string en anglais) | `"hello"`, `'Python'` | ❌ Non |
| `NoneType` | Valeur nulle / absence de valeur | `None` | ❌ Non |


````python
# Rappel : vérifier le type d'une variable
age = 25
prix = 19.99
nom = "TheLook"
validation = True
valeur_manquante = None

for var_name, var in [(
    "age", age), 
    ("prix", prix), 
    ("nom", nom),
    ("validation", validation), 
    ("valeur_manquante", valeur_manquante)]:

    print(f"{var_name:20s} = {str(var):10s}  →  type : {type(var).__name__}")
````

````python
# Conversions de types (casting)
print(int(3.7))          # 3  — troncature, pas arrondi !
print(round(3.7))        # 4  - ici on a un arrondi à l'entier supérieur
print(float("42"))       # 42.0
print(str(3.14))         # '3.14'
print(bool(0))           # False — 0, "", [], None → Toute ces valeurs sont dites "Falsy"
print(bool(42))          # True  — tout le reste est considéré comme True

print(bool(" "))        # et ça ?
````

### Indexation et découpe (slicing)
Comment retrouver un élément d'une liste ainsi que comment la découper 

Rappel: Python est en index 0, cad que le 3e élément s'écrit `list[2]`

````python
# Avant de lancer la cellule, devinez ce que va afficher le print  

print(prix_produits[0])      # premier élément
print(prix_produits[-1])     # dernier élément
print(categories[1:4])    #  slice (début inclus, fin exclue)
print(categories[::2])    # un élément sur deux
````

### Mutabilité : on peut modifier une liste sans faire de copie


````python
print(f"list avant modification: {prix_produits}")
prix_produits[1] = 39.99          # remplacement d'un élément à la position 1 de l'index

# function de modification d'une list
prix_produits.append(19.99)       # ajout à la fin
prix_produits.insert(0, 9.99)     # insertion au début car index 0
print(f"list apres insertion{prix_produits}")

prix_produits.pop()               # retire le dernier
prix_produits.sort()              # tri en place (croissant)

print(f"list apres modification: {prix_produits}")
````

### Example de pattern: Ingestion JSON d'une API

- création d'une liste vide
- parcourir les éléments du dictionnaire avec une `For` loop
- utiliser la methode .append() pour ajouter les éléments dans la liste

````python
# Pour l'example: données brutes reçues depuis une API (format JSON / Liste de dictionnaires)
api_users = [
    {"id": 1, "prenom": "alice", "name": "dupont", "est_actif": True},
    {"id": 2, "prenom": "bob", "name": "martin", "est_actif": False},
    {"id": 3, "prenom": "charlie", "name": "durand", "est_actif": True}
]

print(type(api_users))
````

````python

# Étape 1 : Créer une liste vide
users_formated = []

# Étape 2 : Parcourir la réponse JSON avec une boucle for
for user in api_users:

    # Étape 3 : Construire le nouvel objet et l'ajouter avec .append()
    users_formated.append({
        "id": user["id"],
        "name": "name",
        "statut": "Actif" if user["est_actif"] else "Inactif"
    })

print(users_formated)
print(type(users_formated))     # liste de dictionnaire qui pourrait être ingérer dans une base de données 
````

## 5.3 Les limites...

Jusque-là, rien de nouveau. Vous connaissez les listes. Mais observez ce qui se passe quand on essaie de les utiliser pour faire du **calcul sur des données** :

````python
# ❌ PROBLÈME 1 : Les opérateurs arithmétiques ne font PAS ce qu'on attend

prix = [29.99, 49.99, 14.50, 89.00, 34.99]

# On voudrait appliquer -20% à tous les prix…
# prix * 0.8   → TypeError !

# Ce qui se passe avec * sur une liste :
print([1, 2, 3] * 3)
# [1, 2, 3, 1, 2, 3, 1, 2, 3]  — c'est de la RÉPÉTITION, pas une multiplication !

# Et l'addition ?
print([1, 2, 3] + [4, 5, 6])
# [1, 2, 3, 4, 5, 6]  — c'est de la CONCATÉNATION, pas une addition de chaque élément!
````

````python
# ❌ PROBLÈME 2 : Pour calculer, on est obligé de faire une boucle ou une fonction manuelles

prix = [29.99, 49.99, 14.50, 89.00, 34.99]

# Appliquer -20% il faut une liste comprehension
prix_reduits = [p * 0.8 for p in prix]

# Equivalent avec une For loop:
p_r = []
for p in prix:
    p_r.append(p * 0.8)

print(f"Prix réduits : {prix_reduits}")

# Calculer la moyenne → il faut une formule manuelle
moyenne = sum(prix) / len(prix)
print(f"Moyenne : {moyenne:.2f}")

# Ça marche… mais c'est VERBEUX et LENT sur de grandes données.
````

````python
# ❌ PROBLÈME 3 : Les performances s'effondrent avec les gros volumes

import time

# Créons une liste d'1 million de prix
n = 1_000_000
gros_prix = list(range(n))

# Mesurer le temps pour calculer la somme avec une boucle
debut = time.perf_counter()
total = 0
for p in gros_prix:
    total += p
temps_boucle = time.perf_counter() - debut

# Mesurer avec sum() (C sous le capot, mais toujours une liste Python)
debut = time.perf_counter()
total = sum(gros_prix)
temps_sum = time.perf_counter() - debut

print(f"Boucle for : {temps_boucle*1000:.1f} ms")
print(f"sum()      : {temps_sum*1000:.1f} ms")
print(f"\n Et avec NumPy ? On verra dans le prochain notebook...")
````

### 🤔 Pourquoi les listes sont-elles lentes ?

C'est une question d'**architecture mémoire**. Chaque élément d'une liste Python est un **objet complet** avec son propre type, son compteur de références, et sa valeur. La liste elle-même ne contient que des **pointeurs** vers ces objets, dispersés en mémoire.


Pour chaque opération, Python doit :
1. Suivre le pointeur
2. Vérifier le type de l'objet
3. Extraire la valeur
4. Faire le calcul
5. Créer un nouvel objet résultat

**× 1 million de fois.**

C'est là que **NumPy** entre en jeu.

## 5.4 De la Liste au NumPy Array — Un Aperçu

NumPy stocke les données de manière **complètement différente** : un bloc **contigu** de valeurs du **même type**, directement en mémoire, sans pointeurs ni objets Python intermédiaires.

```
NumPy array :     [ 29.99 | 49.99 | 14.50 | 89.00 | 34.99 ]
                   ←────── bloc contigu en mémoire ──────→
                   (même type : float64 · cache CPU efficace)
```

Résultat : les opérations mathématiques sont envoyées au processeur **en une seule instruction**, sans boucle Python. C'est **10 à 100× plus rapide**.

Voyons la différence :

````python
import numpy as np

prix_liste = [29.99, 49.99, 14.50, 89.00, 34.99]
prix_array = np.array(prix_liste)       # conversion de la liste en 1D array

print(f"Liste Python : {prix_liste}")
print(f"  type = {type(prix_liste)}")
print()
print(f"NumPy array  : {prix_array}")
print(f"  type  = {type(prix_array)}")
print(f"  dtype = {prix_array.dtype}")  # ← tous les éléments ont le MÊME type
````

````python
# Avec NumPy, les opérateurs font ce qu'on attend :

print("Réduction -20% :", prix_array * 0.8)       # multiplication élément par élément
print("TVA +20%       :", prix_array * 1.2)
print("Arrondi        :", np.round(prix_array * 0.8, 2))
print()
print("Addition       :", np.array([1, 2, 3]) + np.array([10, 20, 30]))  # → [11, 22, 33]
print()

# Comparez avec les listes :
# [1, 2, 3] + [10, 20, 30]  → [1, 2, 3, 10, 20, 30]  ← concaténation
# np.array([1,2,3]) + np.array([10,20,30]) → [11, 22, 33] ← addition
````

````python
# Statistiques en UNE ligne :

print(f"Moyenne : {prix_array.mean():.2f} €")
print(f"Médiane : {np.median(prix_array):.2f} €")
print(f"Écart-type : {prix_array.std():.2f} €")
print(f"Min / Max : {prix_array.min():.2f} / {prix_array.max():.2f} €")
````

````python
#  Performance : le moment de vérité !

n = 1_000_000
gros_prix_liste = list(range(n))
gros_prix_array = np.arange(n)

# Boucle Python
debut = time.perf_counter()
total_py = sum(gros_prix_liste)
temps_py = time.perf_counter() - debut

# NumPy
debut = time.perf_counter()
total_np = np.sum(gros_prix_array)
temps_np = time.perf_counter() - debut

print(f"Python sum()  : {temps_py*1000:.2f} ms")
print(f"NumPy sum()  : {temps_np*1000:.2f} ms")
print(f"\n NumPy est bien plus rapide !")
````

### Résumé : Liste vs. NumPy Array

| | Liste Python `list` | NumPy `ndarray` |
| --- | --- | --- |
| **Types** | Hétérogènes (mix `int`, `str`…) | Homogènes (un seul `dtype`) |
| **Mémoire** | Pointeurs vers des objets dispersés | Bloc contigu de valeurs |
| **`*` opérateur** | Répétition (`[1,2] * 3` → `[1,2,1,2,1,2]`) | Multiplication (`arr * 3` → `[3,6]`) |
| **`+` opérateur** | Concaténation | Addition élément par élément |
| **Filtrage** | `[x for x in l if x > 5]` (boucle) | `arr[arr > 5]` (masque) |
| **Statistiques** | `sum(l)/len(l)` (manuel) | `arr.mean()` (intégré) |
| **Performance** | ~100 ms pour 1M éléments | ~1 ms pour 1M éléments |
