## 4. Manipuler les dates avec .dt


### Extraire les composantes d'une date
Par défaut, lors de la lecture d'un fichier (ex: CSV), Pandas importe les dates sous forme de simples chaînes de caractères (object ou str).

Sous forme de texte, vous ne pouvez pas:
- effectuer de calculs temporels 
- extraire facilement des composants de date (example: jour, mois etc...)

Une fois la colonne convertie en datetime, on peut en extraire plein d'informations.

```python
# Chaque composante devient une nouvelle colonne utilisable pour l'analyse

orders['annee'] = orders['created_at'].dt.year
orders['mois'] = orders['created_at'].dt.month
orders['jour_semaine'] = orders['created_at'].dt.day_name()
orders.head()
```

```python
# Vérifiez le type des nouvelles colonnes créées
orders.info()
```

### Calculer un écart entre deux dates

```python
# Rappel de conversion str -> datetime type
orders['shipped_at'] = pd.to_datetime(orders['shipped_at'])

# Le résultat est un timedelta, .dt.days donne le nombre de jours
# on peut utiliser de simple operator comme '-' ou '+'
orders['delai'] = (orders['shipped_at'] - orders['created_at']).dt.days

orders[['created_at', 'shipped_at', 'delai']].head()
```

# Quel est le délai de livraison moyen ? Testez avec .mean()

```python
delai_moyen = orders['delai']."METHODE"()
print(delai_moyen)
```
