## Étape 1.5 : Nettoyer les données
Votre exploration a révélé des problèmes. Avant de faire des statistiques, il faut s'assurer que les données sont propres.

```python
# Traitez les valeurs manquantes (imputez ou supprimez en justifiant votre choix en commentaire)
sales_clean = sales.copy()
# Suppression des lignes avec des valeurs manquantes
sales_clean = sales_clean.dropna()
```

```python
sales_clean.isna().sum()  # Vérification des valeurs manquantes après le nettoyage
```

```python
# Vérifiez et supprimez les doublons exacts s'il y en a
sales_clean = sales_clean.drop_duplicates()
```

```python
# verification du nombre de lignes et colonnes après le nettoyage
sales_clean.shape
```

```python
# Standardisez une colonne texte (ex: mettez les noms de pays en minuscules et sans espaces autour pour vérifier la propreté)
sales_clean['country'] = sales_clean['country'].str.strip().str.lower()
```

```python
# j'affiche les valeurs de la colonne 'country' pour vérifier la standardisation
sales_clean['country']
```

## Étape 2 : Filtrer
Votre manager s'intéresse uniquement au département **Women** et aux ventes de l'année la plus récente.

```python
# Convertissez created_at en datetime
sales_clean['created_at'] = pd.to_datetime(sales_clean['created_at'], errors='coerce')
```

```python
# Trouvez l'année la plus récente
latest_year = sales_clean['created_at'].dt.year.max()
latest_year
```

```python
# Appliquez le double filtre (women department & annee la plus récente)
sales_filtered = sales_clean[(sales_clean['department'] == 'Women') & (sales_clean['created_at'].dt.year == latest_year)]
sales_filtered 
```

## Étape 3 : Transformer
Sur le sous-ensemble filtré, créez les colonnes suivantes :

```python
sales_filtered.dtypes  # Vérification des types de colonnes après le filtrage
```

```python
# Créez 'marge_pct' = (sale_price - cost) / sale_price * 100
sales_filtered['marge_pct'] = (sales_filtered['sale_price'] - sales_filtered['cost'].str.replace('$', '').astype(float)) / sales_filtered['sale_price'] * 100

sales_filtered 
```

```python
# Extrayez le mois de created_at dans une colonne 'mois'
sales_filtered['mois'] = sales_filtered['created_at'].dt.month

sales_filtered['mois']
```

```python
# Créez 'categorie_prix' : 'Budget' (<25), 'Mid-range' (25-75), 'Premium' (>75)
sales_filtered['categorie_prix'] = pd.cut(sales_filtered['sale_price'], bins=[0, 25, 75, float('inf')], labels=['Budget', 'Mid-range', 'Premium'])
sales_filtered['categorie_prix']
```
