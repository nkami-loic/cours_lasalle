## 2. Series et DataFrame

Un **DataFrame** est un tableau à deux dimensions. Chaque colonne d'un dataframe est une **Series** (un tableau 1D avec un index).

```python
# Création d'une Series
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(f"Series: {s}")
```

```python
# convertissons cette Serie en dataframe
df = pd.DataFrame(s)
df

# Que constatez vous?
```

## 4. Sélection avec .loc et .iloc

- `.loc[lignes, colonnes]` : Sélection basée sur les **labels** (les noms des index et colonnes).
- `.iloc[lignes, colonnes]` : Sélection basée sur la **position** entière (comme dans NumPy).

```python
# .iloc (position)
print("Première ligne, deux premières colonnes (iloc) :")
display(products.iloc[0, :2])
```

```python
# Sélection conditionnelle (très courant)
mask = products['retail_price'] > 100
produits_chers = products[mask]
print(f"\nNombre de produits à plus de 100€ : {len(produits_chers)}")
```

### RAPPEL: Quelle est la différence entre = et == dans Python?

```python
# Multiples conditions avec & (ET) et | (OU)
# Les parenthèses sont obligatoires ! 
jeans_chers = products[(products['category'] == 'Jeans') & (products['retail_price'] > 50)]

# ici nous avons pu filtrer avec une double conditions:
print(f"Nombre de jeans à plus de 50€ : {len(jeans_chers)}")
```
