#  Activité  : Audit Post-Mortem du Monolithe FastMarket

---

### 🎯 Votre Mission :
Derrière chaque plainte métier se cache une **erreur d'architecture technique fondamentale**.  
Analysez les 4 symptômes et complétez la **Grille d'Audit Technique** ci-dessous pour identifier la cause racine (*Root Cause*) et le goulot d'étranglement (*Bottleneck*).

---

###  Grille d'Audit à Compléter

| Symptôme Métier | Cause Racine Technique (*Ce qui se passe vraiment sous le capot*) | Composant Saturé (*CPU, RAM, Disque I/O, Locks...*) |
|---|---|---|
| **1. Site gelé à 9h00** *(Rapports BI vs Paniers)* | *Pourquoi une requête de lecture bloque-t-elle des écritures de paniers ?* |    |
| **2. Timeout > 90 jours** *(Requêtes historiques)* |  *Que fait le moteur de base de données face à 10 millions de lignes sans découpage ?* |     |
| **3. Échec Droit à l'oubli** *(Conformité RGPD)* |  *Pourquoi ne peut-on pas juste faire `DELETE FROM users WHERE id = X` ?* |    |
| **4. Factures sans résultats** *(Scalabilité verticale)* |   *Pourquoi doubler la RAM de la machine ne résout rien ?* |    |

---

### 💡 Concepts Clés & Indices
Pour vous guider dans vos analyses, mobilisez les concepts suivants :
* **Conflits de Verrous (*Table Lock / Row Lock*)**
* **Incompatibilité OLTP vs OLAP**
* **Parcours de table complet (*Full Table Scan*)**
* **Clés étrangères & Intégrité référentielle**
* **Limites de la scalabilité verticale (*Scale-up*)**
